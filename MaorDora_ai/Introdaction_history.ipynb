### The Bearth and Fundantimantals

Before the commercial computer, there were many ideas that want to copy the human brain.
* In 1943 Warren McCulloch and Walter Pitts wrote a paper on "A Logical Calculus of the Ideas Immanent in Nervous Activity" that proposed the first mathematical model for building a neural network. 
* In 1950 Alan Turing published a paper titled "Computing Machinery and Intelligence" in which he proposed what is now known as the Turing Test as a criterion of intelligence. 
* In 1956 John McCarthy, Marvin Minsky, Nathaniel Rochester, and Claude Shannon organized the Dartmouth Conference, which is considered the birth of artificial intelligence as a field.Why? Becuase they change they give the field a idenetity and defeintion. 
**"Every aspect of learning or any other feature of intelligence can in principle be so precisely described that a machine can be made to simulate it"**
This mean that they claim that thinking it has no spirit or soul it is a logical proccess that can be copied by machine.
